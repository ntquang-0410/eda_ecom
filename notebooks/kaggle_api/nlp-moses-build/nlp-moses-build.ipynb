```python

import subprocess, os, time
t0 = time.time()
def sh(c, tail=3000, log=None):
    r = subprocess.run(["bash", "-lc", "set -o pipefail; " + c], text=True, capture_output=True)
    out = (r.stdout or "") + (r.stderr or "")
    if log: open(log, "w", encoding="utf-8").write(out)
    print(">>", c[:200], "| ma", r.returncode, "| %.0fs" % (time.time() - t0)); print(out[-tail:], flush=True)
    return r.returncode
os.makedirs("/kaggle/working/moses_pkg", exist_ok=True)
sh("apt-get install -y -qq build-essential cmake git libboost-all-dev zlib1g-dev libbz2-dev liblzma-dev libgoogle-perftools-dev libsparsehash-dev > /tmp/apt.log 2>&1; tail -3 /tmp/apt.log")
sh("git clone --depth 1 https://github.com/moses-smt/mosesdecoder.git /tmp/moses 2>&1 | tail -2; cd /tmp/moses && git log -1 --format=%H")
# Lan 1: bjam mac dinh voi boost he thong
rc = sh("cd /tmp/moses && ./bjam -j4 --with-boost=/usr -q > /kaggle/working/moses_pkg/bjam.log 2>&1; grep -E 'error|Error|failed|\.\.\.' /kaggle/working/moses_pkg/bjam.log | tail -30", tail=6000)
sh("ls /tmp/moses/bin | head -80; ls /tmp/moses/bin | wc -l")
sh("/tmp/moses/bin/moses --version 2>&1 | head -3; /tmp/moses/bin/lmplz --help 2>&1 | head -3; ls /tmp/moses/scripts/training/train-model.perl /tmp/moses/scripts/tokenizer/ | head")
# eflomal
sh("pip install -q eflomal 2>&1 | tail -3; python -c \"import eflomal; print('eflomal', eflomal.__file__)\"")
# dong goi
sh("cd /tmp/moses && tar czf /kaggle/working/moses_pkg/moses_bin_scripts.tgz bin scripts 2>&1 | tail -2; ls -la /kaggle/working/moses_pkg")
print("XONG build", round(time.time() - t0), "giay")
```
