```python
import os, sys, platform, shutil, subprocess, glob

def sh(c):
    r = subprocess.run(c, shell=True, capture_output=True, text=True)
    return (r.stdout + r.stderr).strip()

print("== PYTHON ==", sys.version.replace("\n", " "), "|", sys.executable)
print("platform:", platform.platform(), "| machine:", platform.machine())
print("pip:", sh("pip --version"))
print("uv:", sh("which uv || echo khong co uv"))
print("venv module:", sh("python -c 'import venv, ensurepip; print(\"co venv + ensurepip\")'"))
print("== GPU ==")
print(sh("nvidia-smi --query-gpu=name,memory.total,driver_version,compute_cap --format=csv"))
print(sh("nvcc --version | tail -2 || true"))
print("== GOI DA CAI (loc) ==")
print(sh("pip list 2>/dev/null | grep -i -E '^(torch|torchvision|vllm|transformers|safetensors|tokenizers|huggingface|pandas|pyarrow|numpy|xformers|flash|triton|accelerate|bitsandbytes|autoawq|auto-gptq|optimum|sentencepiece|openai|fastapi|uvicorn|ray|setuptools|wheel|pip) '"))
try:
    import torch
    print("torch:", torch.__version__, "| cuda:", torch.version.cuda, "| is_available:", torch.cuda.is_available(),
          "| so GPU:", torch.cuda.device_count())
    for i in range(torch.cuda.device_count()):
        print(" GPU", i, torch.cuda.get_device_name(i), torch.cuda.get_device_capability(i),
              round(torch.cuda.get_device_properties(i).total_memory / 2**30, 1), "GiB")
except Exception as e:
    print("khong import duoc torch:", type(e).__name__, e)
print("== TAI NGUYEN ==")
print(sh("df -h /kaggle/working /tmp | cat"))
print(sh("free -g | head -2"))
print("so CPU:", os.cpu_count())
print("== /kaggle/input ==")
print(sh("ls -la /kaggle/input | head -30"))
print("== INTERNET ==")
print(sh("timeout 8 curl -sI https://pypi.org | head -1 || echo 'khong co internet (dung nhu mong doi)'"))
```
